```python
import time

t0 = time.time()
mes = ReadSIMEM("055A4D", "2024-01-01", "2024-01-31").main()
t1 = time.time()

print(f"Tiempo: {t1 - t0:.0f} s")
print(f"Filas: {len(mes):,}")
print(f"Memoria: {mes.memory_usage(deep=True).sum() / 1e6:.0f} MB")
```

```python
def procesar_mes(inicio, fin):
    """Descarga un mes de generación real de SiMEM y lo devuelve agregado
    a generación mensual por planta (GWh), usando la versión de liquidación
    más definitiva disponible."""
    # 1. Descargar
    df = ReadSIMEM("055A4D", inicio, fin).main()

    # 2. Versión más definitiva por planta-hora
    df["orden_version"] = df["Version"].map(orden_version)
    df = (
        df.sort_values("orden_version")
          .drop_duplicates(subset=["CodigoPlanta", "FechaHora"], keep="last")
    )

    # 3. Agregar a mensual por planta
    df["mes"] = pd.to_datetime(df["FechaHora"]).dt.to_period("M").astype(str)
    mensual = df.groupby(["CodigoPlanta", "mes"])["Valor"].sum().reset_index()

    # 4. kWh → GWh
    mensual["GWh"] = mensual["Valor"] / 1e6
    return mensual.drop(columns="Valor")
```

```python
ene = procesar_mes("2024-01-01", "2024-01-31")
print(ene.shape)
print(f"Total enero 2024: {ene['GWh'].sum():,.0f} GWh")
```
